```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style='ticks', context='talk')
```

```python
datos = np.random.normal(loc=42, scale=2, size=50)
```

```python
datos[37] = 34
datos[30] = 35
datos[28] = 48
```

```python
mediana = np.median(datos)
```

```python
bp = plt.figure(figsize=(3,6))
sns.boxplot(datos, color='cyan', linecolor='black', saturation=1,
            width=0.7,
            linewidth=1.5,
            fliersize=8, # Tamaño de los outliers
            flierprops={'markerfacecolor':'black'}
           )

ancho_flecha=1
ac = 8
plt.annotate(
    'Pregunta 1',
    xy=(0.25, mediana),  # El punto al que apunta la flecha
    xytext=(0.5, 41.5),  # Dónde se coloca el texto
    fontsize=16,
    arrowprops=dict(
        #facecolor='black',
        color='black',
        shrink=0.05, width=ancho_flecha, headwidth=ac
    ),  # Personalización de la flecha
    va="center",
)

plt.annotate(
    'Pregunta 2',
    xy=(0.25, np.percentile(datos, 25)),  # El punto al que apunta la flecha
    xytext=(0.5, 38),  # Dónde se coloca el texto
    fontsize=16,
    arrowprops=dict(
        color='black', shrink=0.05, width=ancho_flecha, headwidth=ac
    ),  # Personalización de la flecha
    va="center",
)

plt.annotate(
    'Pregunta 3',
    xy=(0.25, np.percentile(datos, 75)),  # El punto al que apunta la flecha
    xytext=(0.5, 44),  # Dónde se coloca el texto
    fontsize=16,
    arrowprops=dict(
        color='black',
        shrink=0.05,
        width=ancho_flecha,
        headwidth=ac
    ),  # Personalización de la flecha
    va="center",
)

plt.annotate(
    'Pregunta 4',
    xy=(0.02, datos[28]),  # El punto al que apunta la flecha
    xytext=(0.5, 47.9),  # Dónde se coloca el texto
    fontsize=16,
    arrowprops=dict(
        color='black', shrink=0.05, width=ancho_flecha, headwidth=ac
    ),  # Personalización de la flecha
    va="center",
)

sns.despine()
plt.show()
```

```python
bp.savefig('quiz-3-boxplot.jpg',
           bbox_inches='tight',
           dpi=200,
           transparent=True
          )
```
