# Joins

Los *joins* sirven para combinar tablas relacionadas usando una o varias columnas en común.

````python
import pandas as pd
````

````python
especies = pd.DataFrame({
    "id_especie": [1, 2, 3, 4],
    "especie": ["Homo sapiens", "Pan troglodytes",
                "Mus musculus", "Danio rerio"]
})
````

````python
especies
````

````python
muestras = pd.DataFrame({
    "id_muestra": [101, 102, 103, 104, 105],
    "id_especie": [1, 2, 1, 3, 99],
    "tejido": ["sangre", "músculo", "hígado", "cerebro", "piel"]
})
````

````python
muestras
````

¿Qué columna podemos usar para relacionar ambas tablas?

## Inner Join

````python
resultado = pd.merge(
    muestras,
    especies,
    on="id_especie",
     )
````

````python
resultado
````

## Left Join

El *left join* conserva todas las filas de la tabla de la izquierda, aunque no tengan correspondencia a la derecha.

````python
muestras
````

````python
especies
````

Si elegimos la tabla `especies` como tabla de la izquierda:

````python
resultado = pd.merge(
    especies,
    muestras,
    on="id_especie",
    how="left"
)
````

````python
resultado
````

Si elegimos la tabla `muestras` como tabla de la izquierda:

````python
resultado = pd.merge(
    muestras,
    especies,
    on='id_especie',
    how='left'
)
````

````python
resultado
````

## Right Join

Es lo contrario del *left join*: conserva todas las filas de la tabla de la derecha.

````python
resultado = pd.merge(
    muestras,
    especies,
    on="id_especie",
    how="right"
)
````

````python
resultado
````

## Full Outer Join

También llamado *full join* u *outer join*.  
Conserva todas las filas de ambas tablas, tengan o no correspondencia.

````python
muestras
````

````python
especies
````

````python
resultado = pd.merge(
    muestras,
    especies,
    on="id_especie",
    how="outer"
)
````

````python
resultado
````

## Join usando nombres de columnas diferentes

No siempre las columnas tienen el mismo nombre.  
Por ejemplo:

````python
especies = pd.DataFrame({
    "codigo": [1, 2, 3],
    "especie": ["Homo sapiens", "Mus musculus", "Danio rerio"]
})

muestras = pd.DataFrame({
    "id_muestra": [101, 102, 103],
    "id_especie": [1, 2, 3]
})
````

````python
especies
````

````python
muestras
````

````python
resultado = pd.merge(
    muestras,
    especies,
    left_on="id_especie",
    right_on="codigo",
    how="inner"
)
````

````python
resultado
````

## Join usando varias columnas

````python
expresión = pd.DataFrame({
    "especie": ["humano", "humano", "ratón", "ratón"],
    "gen": ["TP53", "BRCA1", "Tp53", "Brca1"],
    "expresion": [12.4, 8.7, 15.2, 6.1]
})
````

````python
expresión
````

````python
genes = pd.DataFrame({
    "especie": ["humano", "humano", "ratón", "ratón"],
    "gen": ["TP53", "BRCA1", "Tp53", "Brca1"],
    "funcion": [
        "supresor tumoral",
        "reparación del ADN",
        "supresor tumoral",
        "reparación del ADN"
    ]
})
````

````python
genes
````

````python
resultado = pd.merge(
    expresión,
    genes,
    on=["especie", "gen"],
    how="inner"
)
````

````python
resultado
````
