## Correlación de Pearson

Usando el método `corr` de Pandas:

```python
pearson = datos["longitud_mm"].corr(
    datos["peso_mg"],
    method="pearson"
)

print("Correlación de Pearson:", pearson)
```

El paquete SciPy me permite obtener también el valor p:

```python
from scipy.stats import pearsonr

r, p = pearsonr(datos["longitud_mm"], datos["peso_mg"])

print("Coeficiente r:", r)
print("Valor p:", p)
```

## Correlación de Spearman

Se utiliza para evaluar una relación monótona, incluso cuando no es lineal. Es útil con datos ordinales o cuando la relación entre variables no cumple los supuestos habituales de Pearson.

```python
spearman = datos["longitud_mm"].corr(
    datos["peso_mg"],
    method="spearman"
)

print("Correlación de Spearman:", spearman)
```

Con el paquete SciPy:

```python
from scipy.stats import spearmanr

rho, p = spearmanr(
    datos["longitud_mm"],
    datos["peso_mg"]
)

print("Coeficiente rho:", rho)
print("Valor p:", p)
```

Si tienes varias variables numéricas en un DataFrame de Pandas, puedes calcular la matriz de correlaciones completa:

```python
penguins.corr()
```

```python
# Correlaciones de Pearson
matriz_pearson = penguins.corr(method="pearson", numeric_only=True)

# Correlaciones de Spearman
matriz_spearman = penguins.corr(method="spearman", numeric_only=True)

print("Pearson:")
print(matriz_pearson)

print("\nSpearman:")
print(matriz_spearman)
```

Podemos hacer un mapa de calor (heatmap) para visualizar las correlaciones según una escala de color:

```python
sns.heatmap(matriz_pearson, annot=True,
            vmin=-1, vmax=1, cmap="coolwarm")
plt.show()
```
